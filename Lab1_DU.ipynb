```python
import numpy as np

# 1. Create two sample tensors
tensor_a = np.array([[1, 2, 3], [4, 5, 6]])
tensor_b = np.array([[7, 8, 9], [10, 11, 12]])

# 2. Element-wise addition
addition_result = tensor_a + tensor_b
print("Element-wise Addition:\n", addition_result)

# 3. Element-wise multiplication
multiplication_result = tensor_a * tensor_b
print("\nElement-wise Multiplication:\n", multiplication_result)

# 4. Reshape a tensor
reshaped_tensor = tensor_a.reshape(3, 2)
print("\nReshaped Tensor:\n", reshaped_tensor)

# 5. Slicing and indexing tensors
sliced_tensor = tensor_a[:, 1:]  # Slice second and third columns
print("\nSliced Tensor:\n", sliced_tensor)

# 6. Combining tensors
combined_tensor = np.concatenate((tensor_a, tensor_b), axis=0)  # Stack vertically
print("\nCombined Tensor:\n", combined_tensor)

# 7. Splitting tensors
split_tensors = np.split(combined_tensor, 2)  # Split into two tensors
print("\nSplit Tensors:\n", split_tensors)

# 8. Basic mathematical operations
mean_value = np.mean(tensor_a)
sum_value = np.sum(tensor_b)
print("\nMean of Tensor A:", mean_value)
print("Sum of Tensor B:", sum_value)
```

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

# Define a simple neural network model
model = Sequential([
    Dense(32, activation='relu', input_shape=(10,)),  # Input layer with 10 features
    Dense(16, activation='relu'),                    # Hidden layer
    Dense(1, activation='sigmoid')                   # Output layer
])

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Print model summary
model.summary()
```

```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
import matplotlib.pyplot as plt

# 1. Load and preprocess the dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train = x_train.reshape(-1, 28*28).astype('float32') / 255.0
x_test = x_test.reshape(-1, 28*28).astype('float32') / 255.0
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 2. Define a simple neural network model for classification
model = Sequential([
    Dense(128, activation='relu', input_shape=(784,)),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

# 3. Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# 4. Train the model
history = model.fit(x_train, y_train, epochs=10, validation_split=0.2, batch_size=32)

# 5. Evaluate the model and plot performance indicators
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)

# 6. Plot training and validation loss
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.legend()
plt.title("Loss Over Epochs")
plt.show()

# 7. Plot training and validation accuracy
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.legend()
plt.title("Accuracy Over Epochs")
plt.show()
```

```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences

# 1. Load and preprocess the dataset
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=10000)
x_train = pad_sequences(x_train, maxlen=200)
x_test = pad_sequences(x_test, maxlen=200)

# 2. Define a simple neural network model for classification
model = Sequential([
    Dense(128, activation='relu', input_shape=(200,)),
    Dense(64, activation='relu'),
    Dense(1, activation='sigmoid')
])

# 3. Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 4. Train the model
history = model.fit(x_train, y_train, epochs=5, validation_split=0.2, batch_size=32)

# 5. Evaluate the model and plot performance indicators
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)

# 6. Plot training and validation loss
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.legend()
plt.title("Loss Over Epochs")
plt.show()

# 7. Plot training and validation accuracy
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.legend()
plt.title("Accuracy Over Epochs")
plt.show()
```
