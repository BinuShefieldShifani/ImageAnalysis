```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


image = cv2.imread(cv2.samples.findFile("img.tif")) 

if image is None:
    print("Error: Could not load the image.")
    exit()


m = np.max(image)  
c = 255 / np.log(1 + m)


log_transformed = c * np.log(1 + image.astype(np.float64))

log_transformed = np.uint8(cv2.normalize(log_transformed, None, 0, 255, cv2.NORM_MINMAX))

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(image, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Log Transformed Image')
plt.imshow(log_transformed, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread(cv2.samples.findFile("img.tif")) 

if image is None:
    print(f"Error: Could not load the image from path '{image_path}'. Please check the file path.")
    exit()


normalized_image = image / 255.0


gamma_values = [0.5, 2.0] 
gamma_corrected_images = []

for gamma in gamma_values:
    gamma_corrected = np.array(255 * (normalized_image ** gamma), dtype=np.uint8)
    gamma_corrected_images.append(gamma_corrected)


plt.figure(figsize=(15, 5))


plt.subplot(1, 3, 1)
plt.title('Original Image')
plt.imshow(image, cmap='gray')
plt.axis('off')

# Gamma < 1
plt.subplot(1, 3, 2)
plt.title(f'Gamma < 1 (Gamma={gamma_values[0]})')
plt.imshow(gamma_corrected_images[0], cmap='gray')
plt.axis('off')

# Gamma > 1
plt.subplot(1, 3, 3)
plt.title(f'Gamma > 1 (Gamma={gamma_values[1]})')
plt.imshow(gamma_corrected_images[1], cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread(cv2.samples.findFile("gaussianNoiseImg.tif")) 

if image is None:
    print(f"Error: Could not load the image from path '{image_path}'. Please check the file path.")
    exit()

kernel_size_5x5 = (5, 5)  
gaussian_blurred_5x5 = cv2.GaussianBlur(image, kernel_size_5x5, sigmaX=0)


plt.figure(figsize=(15, 5))


plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')


plt.subplot(1, 2, 2)
plt.title('Gaussian Blur with 5x5 Kernel')
plt.imshow(cv2.cvtColor(gaussian_blurred_5x5, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.tight_layout()
plt.show()

```
